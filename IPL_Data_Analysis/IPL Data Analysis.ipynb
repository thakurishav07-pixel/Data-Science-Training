DATA CLEANING

```python
import numpy as np
import pandas as pd
df=pd.read_csv("IPL.csv")
df.head()
df.columns
df.info()
```

TOP RUN SCORERS

```python
top_runs = df.groupby("batter")["runs_batter"].sum()
top_runs = top_runs.sort_values(ascending=False)
print(top_runs.head(10))
```

MOST WICKETS

```python
top_wickets = df.groupby("bowler")["bowler_wicket"].sum()
top_wickets = top_wickets.sort_values(ascending=False)
print(top_wickets.head(10))
```

LEAST RUN SCORERS

```python
least_runs = df.groupby("batter")["runs_batter"].sum()
least_runs = least_runs.sort_values(ascending=True)
print(least_runs.head(10))
```

TEAM WINNING PERCENTAGE

```python
wins = df["match_won_by"].value_counts()
print(wins)
print()
win_percent = (wins / wins.sum())*100
print(win_percent)
```

TOSS IMPACT ANALYSIS

```python
toss_match_win = df[df["toss_winner"]==df["match_won_by"]]
percentage = (len(toss_match_win) / len(df))*100
print("Toss Impact Percentage = ",percentage)
```
